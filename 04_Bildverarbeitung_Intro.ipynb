```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

# Bilder einlesen

```python
miata = Image.open("./auto_bilder/color.png")
miata
```

```python
plt.imshow(miata)

print(miata.format, miata.size, miata.mode)
```

```python
miata_np = np.asarray(miata)
print(miata_np.dtype)
```

# Bilder croppen & resizen

```python
resized_np = miata_np[::4,::4]
plt.imshow(resized_np)
```

```python
resized_miata = miata.resize((275//4, 183//4))
resized_miata
```

```python
fig, ax = plt.subplots(2,2)
ax[0,0].imshow(miata_np)
ax[1,0].imshow(resized_np)
ax[1,1].imshow(resized_miata)
```

```python
tire = miata.crop((60,100,120,180))
plt.imshow(tire)
```

Erzeuge eine Matrix M (5x4) mit zufälligen Integerwerten 0-300 und dem Datentyp uint16<br>

- Maskiere alle Werte der Matrix M welche größer als 255 sind und sezte diese auf den Wert 255. Speichere das Ergebnis in einer neuen Matrix mit dem Datentyp **uint8**. Schaue in der Doku mach mit welchem Befehl sich der Datentyp ändern läßt.

- Gib alle Elemente der Matrix M aus deren Wert eine 'gerade' Zahl ist.


```python
M = np.random.randint(low=0, high=300, dtype="uint16", size=(5,4))
M
```

```python
M2 = M.copy()
M2[M2 > 255] = 255
M2 = M2.astype("uint8")
M2
```

```python
M[M % 2 == 0]
```

Erzeuge 2 Matrizen M, N . Beide sollen die Dimension 2x2 haben und mit zufälligen Integerwerten befüllt sein.

- Erzeuge eine neue Matrix X indem die beiden Matrizen M, N horizontal aneinandergefügt werden. Gib die Werte der Matrix X sowie die Shape aus
- Erzeuge eine neue Matrix X indem die beiden Matrizen M, N vertikal aneinandergefügt werden. Gib die Werte der Matrix X sowie die Shape aus

```python
M,N = [np.random.randint(low=0, high=255,size=(2,2)) for _ in range(2)]
```

```python
X = np.hstack((M,N))
print(X.shape)
X
```

```python
X = np.vstack((M,N))
print(X.shape)
X
```

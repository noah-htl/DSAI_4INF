```python
import numpy as np
from typing import Any

from cat_record import CatRecord
```

```python
def output_info(np_array: np.ndarray):
    print(f"Shape: {np_array.shape}")
    print(f"NDim: {np_array.ndim}")
    print(f"DType: {np_array.dtype}")
    print(f"Size: {np_array.size}")
    print(f"Content:\n{np_array}")
```

```python
a = np.array([1,2,3,4,5,6])
output_info(a)
```

```python
b = np.array(10)
output_info(b)
```

```python
class StudentRecots:
    def __init__(self, name, age: int):
        self.name = name
        self.age = age
    def __str__(self):
        return f"{self.name} ({self.age})"
```

```python
sr = StudentRecots("Emilio", 5)
print(sr)
```

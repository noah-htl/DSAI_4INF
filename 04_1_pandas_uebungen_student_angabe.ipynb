Erzeuge eine neue Spalte mit dem BMI (BodyMassIndex) Wert indem du folgende Formel anwendest:

$ BMI = \dfrac{Weight[kg]}{Height[m]^2} $

Der BMI soll mit einer Kommastelle angegeben werden

````python
df["BMI"] = df.apply(lambda x: round(x.weight / ((x.height / 100) ** 2), 1), axis=1)
````

````python
df
````

Selektiere alle Datensätze deren BMI > 25 ist und verringere deren Einkommen um 1000 Euro.

````python
df.salary = df.apply(lambda x: x.salary-1000 if x.BMI > 25 else x.salary, axis=1)
````

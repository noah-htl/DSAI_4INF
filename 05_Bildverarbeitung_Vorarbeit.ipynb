# Bildverarbeitung

```python
from PIL import Image
import os
from pathlib import Path
import shutil
```

```python
new_path = Path("./bilder")
raw_path = Path("./rohe_bilder")
shutil.rmtree(new_path)
os.mkdir(new_path)
for label in os.listdir(raw_path):
    os.mkdir(new_path / label)
```

```python
for label in os.listdir(raw_path):
    p = raw_path / label
    t = new_path / label
    for i,file in enumerate(os.listdir(p)):
        img = Image.open(p / file)
        img.thumbnail((256,256), Image.Resampling.LANCZOS)
        img.save(t / f"{i}.jpg")
```

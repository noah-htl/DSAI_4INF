```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
```

```python
url = "https://quotes.toscrape.com/"
request = requests.get(url)
request
```

```python
html = BeautifulSoup(request.content, "html.parser")
```

```python
quotes_html = html.find("span", {"class": "text"})
print(quotes_html.text)
```

```python
quotes = []
for quote_html in html.find_all("span", {"class": "text"}):
    quotes.append(quote_html.text)
quotes
```

```python
authors = []
for author_html in html.find_all("small", {"class": "author"}):
    authors.append(author_html.text)
authors
```

```python
df = pd.DataFrame(quotes, index=authors)
df
```

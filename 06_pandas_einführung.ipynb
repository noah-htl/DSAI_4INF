```python
years = range(2014,2019)
years
```

```python
shop1 = pd.Series([2500,2850,2430,2480,3000], index=years, name="Shop1")
shop2 = pd.Series([2501,2851,2431,2481,3001], index=years, name="Shop2")
shop3 = pd.Series([2502,2852,2432,2482,3002], index=years, name="Shop3")
shop1
```

```python
shops_df = pd.concat([shop1, shop2, shop3], axis=1)
shops_df
```

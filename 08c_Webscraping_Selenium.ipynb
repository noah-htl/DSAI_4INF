```python
from selenium import webdriver
from selenium.webdriver.firefox.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
```

```python
options = webdriver.FirefoxOptions()
```

```python
driver = webdriver.Firefox(options)
wait = WebDriverWait(driver, 10)
url = "https://www.youtube.com/@TomscottGo/videos"
try:
    driver.get(url)

    try:
        cookie_button = driver.find_element(By.XPATH, '//span[contains(text(), "Accept")]')
        cookie_button.click()
    except:
        print("Cookie accept button not found")

    wait.until(EC.visibility_of_element_located((By.ID, "thumbnail")))

    oldheight = None
    height = 0
    while height != oldheight:
        oldheight = height
        height = driver.execute_script("return window.scrollMaxY")
        driver.execute_script("window.scrollBy(0, 99999999);")
        wait.until(EC.url_to_be())
        time.sleep(2)

    time.sleep(2)

    for title_link in driver.find_elements(By.ID, "video-title-link"):
        title = title_link.find_element(By.ID, "video-title")
        url = title_link.get_attribute("href")
        name = title.text

        print(url, name)

finally:
    driver.quit()
```
